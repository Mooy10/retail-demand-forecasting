# Phase 7 - Model Selection And Hybrid Forecast
Advanced features, Phase 7 ML models, model selection, and hybrid backtesting results.


```python
from pathlib import Path
import pandas as pd
ROOT = Path('..').resolve()
processed = ROOT / 'data' / 'processed'
reports = ROOT / 'reports'
```

## Basic vs Advanced Features


```python
basic = pd.read_parquet(processed / 'ml_store_department_features.parquet')
advanced = pd.read_parquet(processed / 'ml_store_department_advanced_features.parquet')
pd.DataFrame({'table': ['basic', 'advanced'], 'rows': [len(basic), len(advanced)], 'columns': [basic.shape[1], advanced.shape[1]]})
```

## Phase 6 vs Phase 7


```python
comparison = pd.read_csv(reports / 'advanced_ml_vs_previous.csv').sort_values('weighted_wape')
comparison[['model', 'weighted_mae', 'weighted_wape', 'weighted_rmsse', 'actual_volume']]
```

## Model Selector Distribution


```python
registry = pd.read_parquet(processed / 'model_selection_registry.parquet')
display(registry['best_model'].value_counts())
display(registry['confidence'].value_counts())
registry.pivot_table(index='store_id', columns='dept_id', values='best_model', aggfunc='first')
```

## Hybrid vs Candidates


```python
hybrid_comp = pd.read_csv(reports / 'hybrid_vs_candidates.csv').sort_values('weighted_wape')
hybrid_comp[['model', 'weighted_wape', 'weighted_rmsse', 'series_count']]
```
