# Phase 8 - Out-of-Sample Model Selection Validation
This notebook reads processed Phase 8 outputs and contrasts in-sample model selection, strict holdout validation, and rolling-origin validation.

````python
from pathlib import Path
import pandas as pd
ROOT = Path('..').resolve()
processed = ROOT / 'data' / 'processed'
reports = ROOT / 'reports'
````

## In-Sample vs Out-of-Sample


````python
insample = pd.read_csv(reports / 'hybrid_vs_candidates.csv').sort_values('weighted_wape')
holdout = pd.read_csv(reports / 'holdout_w3_metrics_summary.csv').sort_values('weighted_wape')
display(insample[['model','weighted_wape','weighted_rmsse']].head(6))
display(holdout[['model','weighted_wape','weighted_rmsse']])
````

## Rolling Evaluation


````python
rolling = pd.read_csv(reports / 'rolling_selector_metrics_summary.csv')
rolling[['model','evaluation_window','weighted_wape','weighted_rmsse','series_count']]
````

## Model Distribution And Stability


````python
holdout_registry = pd.read_parquet(processed / 'model_selection_registry_train_w1_w2.parquet')
rolling_registry = pd.read_parquet(processed / 'rolling_selector_registry.parquet')
transition = pd.read_csv(reports / 'model_transition_matrix.csv', index_col=0)
display(holdout_registry['selected_model'].value_counts())
display(rolling_registry.groupby(['evaluation_window','selected_model']).size())
display(transition)
````

## Confidence vs Future Performance


````python
confidence = pd.read_csv(reports / 'confidence_validation.csv')
confidence
````

## Selector Success And Failure Examples


````python
series_metrics = pd.read_csv(reports / 'holdout_w3_metrics_by_series.csv')
hybrid = series_metrics[series_metrics['model'].eq('hybrid_holdout_w3')][['unique_id','wape','rmsse']]
baseline = series_metrics[series_metrics['model'].eq('seasonal_naive_28')][['unique_id','wape','rmsse']]
comp = hybrid.merge(baseline, on='unique_id', suffixes=('_hybrid','_baseline'))
comp['wape_diff'] = comp['wape_hybrid'] - comp['wape_baseline']
display(comp.sort_values('wape_diff').head(10))
display(comp.sort_values('wape_diff', ascending=False).head(10))
````
