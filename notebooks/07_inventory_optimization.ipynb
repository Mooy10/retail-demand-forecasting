# Phase 9 - Simulated Inventory Optimization
This notebook reads processed outputs from the simulated inventory planning module. It does not use real Walmart inventory.


```python
from pathlib import Path
import pandas as pd
ROOT = Path('..').resolve()
processed = ROOT / 'data' / 'processed'
reports = ROOT / 'reports'
```

## Official Forecast And Inventory Projection


```python
forecast = pd.read_parquet(processed / 'official_forecast_with_uncertainty.parquet')
recommendations = pd.read_parquet(processed / 'inventory_recommendations.parquet')
daily = pd.read_parquet(processed / 'inventory_daily_projection.parquet')
forecast.head()
```

## Safety Stock And Reorder Point


```python
recommendations.groupby(['store_id','scenario'])[['safety_stock','reorder_point','recommended_order_quantity']].mean().round(2).head(20)
```

## Recommended Orders And Stockout Risk


```python
display(recommendations['stockout_risk_level'].value_counts())
recommendations.sort_values('recommended_order_quantity', ascending=False)[['unique_id','store_id','dept_id','scenario','forecast_demand_28d','initial_inventory','recommended_order_quantity','stockout_risk_level']].head(20)
```

## Lean, Base, Conservative Scenarios


```python
recommendations.groupby('scenario')[['initial_inventory','recommended_order_quantity','projected_stockout_units','estimated_total_inventory_cost']].sum().round(2)
```
